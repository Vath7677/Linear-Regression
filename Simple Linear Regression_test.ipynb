```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Provide our Student Data
X = np.array([[15], [25], [35], [45], [55], [65], [75], [85], [90]])
y = np.array([20, 35, 30, 50, 45, 65, 60, 85, 80])

# 2. Create the empty Machine Learning Model
model = LinearRegression()

# 3. 'Train' the model
model.fit(X, y)

# 4. Make a prediction for a new student (80 hours)
new_student = np.array([[80]])
prediction = model.predict(new_student)
print(f"Predicted Score: {prediction[0]:.1f}%")

# ADDED THIS LINE BELOW: Calculate predictions for the trendline
y_pred = model.predict(X)

# 5. Create the Graph
plt.figure(figsize=(8, 5))

# Plot the actual data points
plt.scatter(X, y, color='blue', label='Actual Student Data', s=100, zorder=5)

# Plot the regression line (Now y_pred works perfectly!)
plt.plot(X, y_pred, color='red', linewidth=2, label='Linear Regression Line')

# Highlight the new prediction point on the graph
plt.scatter(new_student, prediction, color='green', marker='*', s=200, label='Prediction (80 hrs)', zorder=6)

# Labels and Styling
plt.title('Hours Studied vs. Test Score', fontsize=14, pad=15)
plt.xlabel('Hours Studied', fontsize=12)
plt.ylabel('Test Score (%)', fontsize=12)
plt.xlim(10, 100)
plt.ylim(10, 100)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)

# Display the plot
plt.show()
```
